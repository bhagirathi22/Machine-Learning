```python
# STEP 1 - IMPORT LIBRARIES
# ==========================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
#STEP 2 - LOAD DATASET
# ==========================================================

df = pd.read_csv("Iris.csv")

print("=" * 60)
print("FIRST FIVE RECORDS")
print("=" * 60)
print(df.head())

```

```python
#STEP 3 - LAST FIVE RECORDS
# ==========================================================

print("\nLAST FIVE RECORDS")
print(df.tail())
```

```python
# STEP 4 - SHAPE
# ==========================================================

print("\nDATASET SHAPE")
print(df.shape)

rows, cols = df.shape

print(f"Rows : {rows}")
print(f"Columns : {cols}")

```

```python
# STEP 5 - COLUMN NAMES
# ==========================================================

print("\nCOLUMN NAMES")
print(df.columns)
```

```python
# STEP 6 - DATA TYPES
# ==========================================================

print("\nDATA TYPES")
print(df.dtypes)
```

```python
# STEP 7 - DATASET INFORMATION
# ==========================================================

print("\nDATASET INFORMATION")
df.info()
```

```python
# STEP 8 - STATISTICAL SUMMARY
# ==========================================================

print("\nSTATISTICAL SUMMARY")
print(df.describe())
```

```python
# STEP 9 - MISSING VALUES
# ==========================================================

print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python
# STEP 10 - DUPLICATE VALUES
# ==========================================================

print("\nDUPLICATE VALUES")
print(df.duplicated().sum())
```

```python
# STEP 11 - UNIQUE CLASSES
# ==========================================================

print("\nUNIQUE SPECIES")
print(df["Species"].unique())

print("\nCLASS COUNT")
print(df["Species"].value_counts())
```

```python
# STEP 12 - FEATURE SELECTION
# ==========================================================

X = df.drop(columns=["Species"])

print("\nFEATURES")
print(X.head())
```

```python
# STEP 13 - TARGET VARIABLE
# ==========================================================

y = df["Species"]

print("\nTARGET")
print(y.head())

```

```python
# STEP 14 - TRAIN TEST SPLIT
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
test_size=0.20,
random_state=42,
    stratify=y
)

```

```python
# STEP 15 - DISPLAY SPLIT INFORMATION
# ==========================================================

print("\nTRAIN TEST SPLIT")

print(f"Training Features : {X_train.shape}")
print(f"Testing Features  : {X_test.shape}")

print(f"Training Labels   : {y_train.shape}")
print(f"Testing Labels    : {y_test.shape}")
```

```python
# STEP 16 - DISPLAY SAMPLE TRAINING DATA
# ==========================================================

print("\nTRAINING FEATURES")
print(X_train.head())

print("\nTRAINING LABELS")
print(y_train.head())
```

```python
# STEP 17 - DISPLAY SAMPLE TESTING DATA
# ==========================================================

print("\nTEST FEATURES")
print(X_test.head())

print("\nTEST LABELS")
print(y_test.head())
```

```python
# STEP 18 - CLASS DISTRIBUTION CHECK
# ==========================================================

print("\nTRAINING CLASS DISTRIBUTION")
print(y_train.value_counts())

print("\nTESTING CLASS DISTRIBUTION")
print(y_test.value_counts())
```

```python
# STEP 19 - SUMMARY
# ==========================================================

print("\n" + "=" * 60)
print("MACHINE LEARNING DATA PREPARATION COMPLETED SUCCESSFULLY!")
print("=" * 60)
print("✔ Dataset Loaded")
print("✔ Dataset Explored")
print("✔ Missing Values Checked")
print("✔ Duplicate Values Checked")
print("✔ Features Selected")
print("✔ Target Selected")
print("✔ Train-Test Split Completed")
print("=" * 60)
```
